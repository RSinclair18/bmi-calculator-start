### Problem 4.1: Run a 2-qubit Grover's Algorithm in Cirq to find the marked state |11⟩.

Two qubits can hold four states: 00, 01, 10 and 11. One of them, |11⟩, is the marked state. Run each cell from top to bottom: you'll create the qubits, put them in an equal superposition, apply the oracle and the diffuser, then measure and simulate. Watch the circuit grow with each step, and check that the histogram finds 11.

All of the code is already provided. You just need to run each step.

**Step 1: Create the qubits** — Import Cirq and create two qubits named `q0` and `q1`.

````python
# Import the Cirq library

# Create qubits 'q0' and 'q1' and store them in the variable qubits
qubits = cirq.NamedQubit.range(2, prefix = "q")

# Print the qubits to see their names
print(qubits)
````

**Step 2: Prepare an equal superposition** — Create a circuit and put an `H` gate on every qubit.

````python
# Create an empty circuit
circuit = cirq.Circuit()

# Apply an H gate to each qubit -> equal superposition of all 4 states
circuit.append(cirq.H.on_each(qubits))
````

**Step 3: Apply the oracle** — Flip the sign of the marked state `|11⟩`. An `H` – `CNOT` – `H` sandwich on `q1` does exactly that.

````python
# Oracle: flip the sign of |11>
circuit.append(cirq.H(qubits[1]))
circuit.append(cirq.CNOT(qubits[0], qubits[1]))
circuit.append(cirq.H(qubits[1]))
````

**Step 4: Apply the diffuser** — Flip every amplitude over the average: `H` and `X` on every qubit, the same sign flip as the oracle, then `X` and `H` again.

````python
# Diffuser: flip every amplitude over the average
circuit.append(cirq.H.on_each(qubits))
circuit.append(cirq.X.on_each(qubits))
circuit.append(cirq.H(qubits[1]))
circuit.append(cirq.CNOT(qubits[0], qubits[1]))
circuit.append(cirq.H(qubits[1]))
circuit.append(cirq.X.on_each(qubits))
circuit.append(cirq.H.on_each(qubits))
````

**Step 5: Measure** — Add a measurement to every qubit.

````python
# Append a measurement to all qubits
circuit.append(cirq.measure(qubits))
````

**Step 6: Run the simulator** — Run the circuit 100 times and plot the results.

````python
# Create a quantum simulator
sim = cirq.Simulator()

# Run the circuit 100 times and store the result
result = sim.run(circuit, repetitions=100)

# Visualize the results with a histogram
hist = cirq.plot_state_histogram(result, plt.subplot(), title = 'Qubit States',
                                 xlabel = 'States', ylabel = 'Occurrences',
                                 tick_label=['00','01','10','11'])

plt.show()
````
